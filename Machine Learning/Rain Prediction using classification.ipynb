```python
year = pd.to_datetime(df.Date).dt.year
```

```python
train_df = df[year<2015]
```

```python
inputs = list(train_df.columns)[1:-1]
```

```python
targets = 'RainTomorrow'
```

```python
df.columns
```

```python
train_inputs = train_df[inputs].copy()
train_targets = train_df[targets].copy()
```

```python
val_inputs = val_df[inputs].copy()
val_targets = val_df[targets].copy()
```

```python
test_inputs = test_df[inputs].copy()
test_targets = test_df[targets].copy()
```

```python
train_inputs.head()
```

```python
numerical_cols = train_inputs.select_dtypes(include=np.number).columns.to_list()
categorical_cols = train_inputs.select_dtypes(include='object').columns.to_list()
```

## imputation

```python
from sklearn.impute import SimpleImputer
```

```python
imp = SimpleImputer(strategy='mean').fit(train_inputs[numerical_cols])
```

```python
train_inputs[numerical_cols] = imp.transform(train_inputs[numerical_cols])
```

```python
val_inputs[numerical_cols]=  imp.transform(val_inputs[numerical_cols])
```

```python
test_inputs[numerical_cols] = imp.transform(test_inputs[numerical_cols])
```

```python
train_inputs[numerical_cols].isnull().sum()
```

```python
val_inputs[numerical_cols].isnull().sum()
```

```python
test_inputs[numerical_cols].isnull().sum()
```

## Scaling

- standard scaler == `mean=0`, `std=1`
- minmaxscaler == `(0,1)`

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler().fit(df[numerical_cols])
```

```python
train_inputs[numerical_cols] = scaler.transform(train_inputs[numerical_cols])
```

```python
val_inputs[numerical_cols] = scaler.transform(val_inputs[numerical_cols])
```

```python
test_inputs[numerical_cols] = scaler.transform(test_inputs[numerical_cols])
```

```python
train_inputs[numerical_cols]
```

```python
val_inputs[numerical_cols]
```

```python
test_inputs[numerical_cols]
```

## considering the categorical_cols

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False,handle_unknown='ignore').fit(df[categorical_cols])
encoded_cols = list(ohe.get_feature_names_out(categorical_cols))
len(encoded_cols)
```

```python
train_inputs[encoded_cols] = ohe.transform(train_inputs[categorical_cols])
```

```python
val_inputs[encoded_cols] = ohe.transform(val_inputs[categorical_cols])
```

```python
test_inputs[encoded_cols] = ohe.transform(test_inputs[categorical_cols])
```

```python
train_inputs[encoded_cols]
```

# concatinating the both numerical amd categorical cols

```python
conc_train = train_inputs[numerical_cols + encoded_cols]
```

```python
conc_val =  val_inputs[numerical_cols + encoded_cols]
```

```python
conc_test = test_inputs[numerical_cols + encoded_cols]
```

## model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression().fit(conc_train,train_targets)
```

```python
weights = pd.DataFrame({
    'features': numerical_cols + encoded_cols,
    'weights': model.coef_.tolist()[0]
})
```

```python
weights.head(10)
```

```python
px.histogram(weights.sort_values(by='weights',ascending=False).head(10),x='weights',y='features')
```

```python
model.intercept_
```

## model predictions

```python
X_train = conc_train
```

```python
X_test = conc_test
```

```python
X_val = conc_val
```

```python
X_train
```

```python
preds=  model.predict(X_val)
```

```python
len(preds)
```

```python
val_targets
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
accuracy_score(preds,val_targets)
```

```python
cf=confusion_matrix(val_targets,preds,normalize='true')
```

```python
cf
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(cf, annot=True, fmt=".2f",cmap='viridis')
plt.xlabel('Prediction')
plt.ylabel('Target')
plt.title('Confusion Matrix')
plt.show()
```
